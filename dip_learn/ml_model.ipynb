```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class Neuron:
    def __init__(self, bias, weights, inputs):
        self.weights = weights
        self.bias = bias
        self.inputs= inputs
        self.output = 0
    
    def forward(self):
        return np.dot(self.inputs, self.weights) + self.bias
    

mySingleNeuron = Neuron(bias=0.5, weights=np.array([0.2, 0.8, -0.5]), inputs=np.array([1.0, 0.5, -1.5]))
print("Neuron output:", mySingleNeuron.forward())
```

```python
class NueralNetwork:
    def __init__(self, input_size=784, hidden_layers=[512,512], output_size=10):
        self.input_size = input_size
        self.hidden_layers = hidden_layers
        self.output_size = output_size
        self.weights = []
        self.biases = []
        
        self.weights.append(0.01 * np.random.randn(self.input_size, self.hidden_layers[0]))
        self.biases.append(np.zeros((1, self.hidden_layers[0])))
        
        for i in range(len(self.hidden_layers)-1):
            self.weights.append(0.01 * np.random.randn(self.hidden_layers[i], self.hidden_layers[i+1]))
            self.biases.append(np.zeros((1, self.hidden_layers[i+1])))
            
        self.weights.append(0.01 * np.random.randn(self.hidden_layers[len(self.hidden_layers)-1], self.output_size))
        self.biases.append(np.zeros((1, self.output_size)))
        
    def forward(self, inputs):
        a = [inputs]
        for i in range(len(self.weights)):
            a.append(np.dot(a[i], self.weights[i]) + self.biases[i])
            
        return a[-1]  
    
myNN = NueralNetwork()
sample_input = np.random.randn(1, 784)
print(myNN.weights)
output = myNN.forward(sample_input)
print("Neural Network output shape:", output.shape)
print(output)          
```
