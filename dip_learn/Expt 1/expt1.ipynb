```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import pathlib

image_path = pathlib.Path(os.path.join(os.getcwd(), "Images"))
```

```python
print(image_path)
```

```python
img_bgr = cv2.imread(image_path / "Nature.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

plt.figure(figsize=(10,12))

plt.subplot(3, 2, 1)
plt.imshow(img_bgr)
plt.title("BGR Image")
plt.axis('off')

plt.subplot(3, 2, 2)
plt.imshow(img_rgb)
plt.title("RGB Image")
plt.axis('off')

plt.subplot(3, 2, 3)
plt.imshow(img_gray) # uses cmap = 'virdis'
plt.title("Grayscale Image (without cmap arg)")
plt.axis('off')

plt.subplot(3, 2, 4)
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image (with cmap='gray' arg)")
plt.axis('off')

plt.subplot(3, 2, 5)
plt.imshow(img_hsv)
plt.title("HSV Image (passed directly to imshow)")
plt.axis('off')

plt.subplot(3, 2, 6)
plt.imshow(cv2.cvtColor(img_hsv, cv2.COLOR_HSV2RGB))
plt.title("HSV Image (converted to RGB)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 18))

plt.subplot(4, 2, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(4, 2, 2)
plt.hist(img_gray.ravel(), 256, (0, 256))
plt.title("Histogram of Grayscale Image before equalization")

img_gray_equalized = cv2.equalizeHist(img_gray)

plt.subplot(4, 2, 3)
plt.imshow(img_gray_equalized, cmap="gray")
plt.title("Equalized Grayscale Image")
plt.axis("off")

plt.subplot(4, 2, 4)
plt.hist(img_gray_equalized.ravel(), 256, (0, 256))
plt.title("Histogram of Grayscale Image after equalization")

r, g, b = cv2.split(img_rgb)
eq_r = cv2.equalizeHist(r)
eq_g = cv2.equalizeHist(g)
eq_b = cv2.equalizeHist(b)
img_rgb_wrongly_equalized = cv2.merge((eq_r, eq_g, eq_b))
plt.subplot(4, 2, 5)
plt.imshow(img_rgb)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(4, 2, 6)
plt.imshow(img_rgb_wrongly_equalized)
plt.title("Wrongly Equalized RGB Image")
plt.axis("off")

h, s, v = cv2.split(img_hsv)
eq_v = cv2.equalizeHist(v)
eq_s = cv2.equalizeHist(s)
eq_hsv2 = cv2.merge((h, s, eq_v)) 
eq_hsv = cv2.merge((h, eq_s, eq_v))
img_rgb_correctly_equalized = cv2.cvtColor(eq_hsv, cv2.COLOR_HSV2RGB)
img_rgb_2 = cv2.cvtColor(eq_hsv2, cv2.COLOR_HSV2RGB)

plt.subplot(4, 2, 7)
plt.imshow(img_rgb_correctly_equalized)
plt.title("Correctly Equalized RGB Image")
plt.axis("off")

plt.subplot(4, 2, 8)
plt.imshow(img_rgb_2)

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite(f'{image_path / 'img_gray.jpg'}', img_gray)
cv2.imwrite(f'{image_path / 'img_gray_equalized.jpg'}', img_gray_equalized)
cv2.imwrite(f'{image_path / 'img_bgr.jpg'}', img_bgr)
cv2.imwrite(f'{image_path / 'img_rgb.jpg'}', cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR))
cv2.imwrite(f'{image_path / 'img_rgb_wrongly_equalized.jpg'}', cv2.cvtColor(img_rgb_wrongly_equalized, cv2.COLOR_RGB2BGR))
cv2.imwrite(f'{image_path / 'img_rgb_correctly_equalized.jpg'}', cv2.cvtColor(img_rgb_correctly_equalized, cv2.COLOR_RGB2BGR))
cv2.imwrite(f'{image_path / 'img_hsv.jpg'}', cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR))
```
